```python
# In thông tin sinh viên theo yêu cầu
print("Họ và tên: Vũ Trương Phi")
print("MSSV: 6451071059")
print("-" * 40)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

# 1. Đọc dữ liệu (Lưu ý: file CSV của tập bank marketing dùng dấu chấm phẩy ';' làm dấu phân cách)
df_bank = pd.read_csv('bank.csv', sep=';')

# 2. Khảo sát thông tin cấu trúc dữ liệu
print("--- THÔNG TIN DỮ LIỆU TẬP BANK MARKETING ---")
df_bank.info()
print("-" * 40)

# 3. Xem 10 dòng đầu tiên của tập dữ liệu
print("--- 10 DÒNG ĐẦU TIÊN CỦA DỮ LIỆU ---")
display(df_bank.head(10))

# 4. Kiểm tra và thống kê dữ liệu khuyết thiếu (Missing values / Null)
print("--- KIỂM TRA GIÁ TRỊ NULL / THIẾU ---")
print(df_bank.isnull().sum())
print("-" * 40)

# 5. Tiền xử lý dữ liệu:
# - Cắt tỉa khoảng trắng thừa ở các cột chuỗi ký tự (object)
obj_cols = df_bank.select_dtypes(include=['object']).columns
for col in obj_cols:
    df_bank[col] = df_bank[col].str.strip()

# - Xóa các dòng trùng lặp (nếu có)
df_bank.drop_duplicates(inplace=True)

# 6. Thống kê và vẽ biểu đồ biến mục tiêu (Khả năng đăng ký tiền gửi có kỳ hạn - cột 'y')
plt.figure(figsize=(6, 4))
df_bank['y'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Thống kê kết quả đăng ký tiền gửi (y)')
plt.xlabel('Đăng ký (yes/no)')
plt.ylabel('Số lượng khách hàng')
plt.xticks(rotation=0)
plt.show()

print("Hoàn thành khảo sát và tiền xử lý dữ liệu Bank Marketing!")
```
